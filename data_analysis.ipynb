```python
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
# Load cleaned data
df = pd.read_csv('sales_data_cleaned.csv')
```

```python
# Convert date column to datetime (just in case)
df['date'] = pd.to_datetime(df['date'])

# Create additional time-based columns
df['month'] = df['date'].dt.to_period('M')
df['year'] = df['date'].dt.year
```

```python
# 1. Total Revenue by Month
monthly_revenue = df.groupby('month')['total_revenue'].sum()
monthly_revenue.plot(kind='line', marker='o')
plt.title('Monthly Revenue')
plt.xlabel('Month')
plt.ylabel('Revenue')
plt.xticks(rotation=45)
plt.tight_layout()
plt.show()
```

```python
# 2. Top 5 Product Categories by Revenue
top_categories = df.groupby('product_category')['total_revenue'].sum().sort_values(ascending=False).head(5)
top_categories.plot(kind='bar', color='skyblue')
plt.title('Top 5 Product Categories by Revenue')
plt.ylabel('Revenue')
plt.xlabel('Product Category')
plt.tight_layout()
plt.show()
```

```python
# 3. Revenue by Payment Method
payment_revenue = df.groupby('payment_method')['total_revenue'].sum().sort_values()
payment_revenue.plot(kind='barh', color='salmon')
plt.title('Revenue by Payment Method')
plt.xlabel('Revenue')
plt.ylabel('Payment Method')
plt.tight_layout()
plt.show()
```

```python
# 4. Delivery Status Counts
delivery_counts = df['delivery_status'].value_counts()
delivery_counts.plot(kind='pie', autopct='%1.1f%%', startangle=90)
plt.title('Delivery Status Distribution')
plt.ylabel('')
plt.tight_layout()
plt.show()
```

```python
# 5. Best-Selling Products (by quantity)
top_products = df.groupby('product_name')['quantity_sold'].sum().sort_values(ascending=False).head(10)
top_products.plot(kind='barh', color='limegreen')
plt.title('Top 10 Best-Selling Products')
plt.xlabel('Total Quantity Sold')
plt.tight_layout()
plt.show()
```
